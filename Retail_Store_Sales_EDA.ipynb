### STEP 4: Data Cleaning
In this step, we clean the dataset by handling missing values, removing duplicate records, correcting data types, fixing inconsistent values, and preparing the data for accurate analysis. This improves the quality and reliability of the dataset.


#### 4.1] Missing Values

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
df["Discount_%"]=df["Discount_%"].fillna(0)
```

```python
df.isnull().sum()
```

#### 4.2] Duplicate Records

```python
df.duplicated()
```

```python
df.duplicated().sum()
```

```python
df=df.drop_duplicates()
```

```python
df.duplicated().sum()
```

#### 4.3] Incorrect DataType

```python
print(df.dtypes)
```

```python
df["Order_Date"]=pd.to_datetime(df["Order_Date"])
```

```python
df.dtypes
```

#### 4.4] Replace Incorrect Values

```python
print("City    : ",df["City"].unique().tolist())
print("Region  : ",df["Region"].unique().tolist())
print("Category: ",df["Category"].unique().tolist())
print("Product : ",df["Product"].unique().tolist())
print("Payment_Method: ",df["Payment_Method"].unique().tolist())
print("Ship_Mode: ",df["Ship_Mode"].unique().tolist())
```

```python
#Remove Extra Space
df["City"]=df["City"].str.strip()
```

```python
print("City    : ",df["City"].unique().tolist())
print("Region  : ",df["Region"].unique().tolist())
print("Payment_Method: ",df["Payment_Method"].unique().tolist())
```

```python
df["City"]=df["City"].replace({
    "pune":"Pune",
    "Dalhi":"Delhi"
})

df["Region"]=df["Region"].replace({
    "north":"North"
})

df["Payment_Method"]=df["Payment_Method"].replace({
    "upi":"UPI","cash":"Cash"
})
```

```python
print("City    : ",df["City"].unique().tolist())
print("Region  : ",df["Region"].unique().tolist())
print("Payment_Method: ",df["Payment_Method"].unique().tolist())
```

### STEP 9: Data Visualization
In this step, we create charts and graphs to present the analysis visually. Visualizations make it easier to identify trends, compare performance, and communicate business insights effectively.

##### 9.1) Sales by Category (Bar Chart)

```python
plt.figure(figsize=(8,5))
df.groupby("Category")["Sales"].sum().plot(kind="bar")
plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.show()
```

##### 9.2) Sales by Region (Bar Chart)

```python
df.groupby("Region")["Sales"].sum().plot(kind="bar")
plt.title("Sales by region")
plt.xlabel("Region")
plt.ylabel("Sales")
plt.show()
```

##### 9.3) Payment Method Distribution (Pie Chart)

```python
df["Payment_Method"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%")
plt.title("Payment Method Distribution")
plt.show()
```

##### 9.4) Sales Distribution (Histogram)

```python
plt.hist(df["Sales"],bins=9,edgecolor="Black")
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

##### 9.5) Sales vs Profit (Scatter Plot)

```python
sns.scatterplot(data=df, x="Sales",y="Profit")
plt.title("Sales vs Profit")
plt.grid()
plt.show()
```

##### 9.6) Correlation Heatmap

```python
plt.figure(figsize=(6,4))
sns.heatmap(
    df[["Quantity","Unit_Price","Sales","Profit"]].corr(),
    annot=True,
    cmap="Blues"
)
plt.title("Correlation Heatmap")
plt.show()
```
